```python
import numpy as np

# Define the function
def f(x):
    return (x + 3) ** 2

# Starting point and learning rate
x = np.array(2.0)
learning_rate = 0.1
iterations = 100

# Gradient Descent
for i in range(iterations):
    gradient = 2 * (x + 3)
    x = x - learning_rate * gradient

print("Local minimum x =", x)
print("Minimum value y =", f(x))
```
